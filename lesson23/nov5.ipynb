```python
import gradio as gr
import plotly.express as px

def create_plots():
    # Group by city to understand aggregate sales
    city_sales = df.groupby('City')['sales'].mean().reset_index().sort_values(by='sales', ascending=False)

    # Create a bar plot of average sales by city
    fig_bar = px.bar(
        city_sales,
        x='City',
        y='sales',
        title='Average Sales by City',
        labels={'sales': 'Average Sales ($)', 'City': 'City'},
        color='sales',
        color_continuous_scale='Viridis'
    )

    # Create a scatter plot representing geographically or individually
    fig_scatter = px.scatter(
        df,
        x='longitude',
        y='latitude',
        size='sales',
        color='sales',
        hover_name='City',
        title='Geographic Distribution of Sales by City Latitude & Longitude',
        labels={'sales': 'Sales', 'latitude': 'Latitude', 'longitude': 'Longitude'},
        color_continuous_scale='Viridis'
    )

    return fig_bar, fig_scatter

with gr.Blocks() as demo:
    gr.Markdown("# 📊 City Sales Performance Dashboard")
    gr.Markdown("This dashboard visualizes how product sales perform across different cities using the Advertising dataset.")

    with gr.Tabs():
        with gr.TabItem("Bar Plot (Average Sales)"):
            plot_bar = gr.Plot()
        with gr.TabItem("Scatter Plot (Geographic View)"):
            plot_scatter = gr.Plot()

    # Load the plots into the interface
    bar_fig, scatter_fig = create_plots()
    demo.load(fn=lambda: (bar_fig, scatter_fig), outputs=[plot_bar, plot_scatter])

demo.launch(inline=True)
```
